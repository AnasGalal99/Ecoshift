```python
X_hist = df['year'].values.reshape(-1, 1)
y_hist = df['annual_co2_emissions'].values
```

```python
model_co2 = LinearRegression()
model_co2.fit(X_hist, y_hist)
```

```python
y_pred_hist = model_co2.predict(X_hist)
```

```python
r2 = r2_score(y_hist, y_pred_hist)
mae = mean_absolute_error(y_hist, y_pred_hist)
rmse = np.sqrt(mean_squared_error(y_hist, y_pred_hist))
```

```python
future_years = np.array(range(2025, 2031)).reshape(-1, 1)
y_future_bau = model_co2.predict(future_years)
```

```python
residuals = y_hist - y_pred_hist
degrees_of_freedom = len(y_hist) - 2
std_error = np.sqrt(np.sum(residuals**2) / degrees_of_freedom)
```

```python
t_val = stats.t.ppf(0.975, df=degrees_of_freedom)
margin_of_error = t_val * std_error
```

```python
mean_actual = y_hist.mean()
norm_mae_error = (mae / mean_actual) * 100
model_accuracy_pct = 100 - norm_mae_error
```

```python
print(f"R2 Score: {r2*100:.1f}%")
print(f"Accuracy: {model_accuracy_pct:.1f}%")
print(f"MAE: {mae:,.0f} ton")
print(f"RMSE: {rmse:,.0f} ton")
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import scipy.stats as stats

df_master = pd.read_csv('Egypt_Master_Clean.csv')
df_egypt = df_master[df_master['year'] <= 2024].dropna(subset=['annual_co2_emissions'])

X_hist = df_egypt[['year']].values
y_hist = df_egypt['annual_co2_emissions'].values

candidate_models = {
    'Linear Regression': LinearRegression(),
    'Ridge Regression': Ridge(),
    'Random Forest Regressor': RandomForestRegressor(n_estimators=100, random_state=42),
    'Gradient Boosting Regressor': GradientBoostingRegressor(n_estimators=100, random_state=42)
}


results = []
trained_models = {}

for name, model in candidate_models.items():
    model.fit(X_hist, y_hist)
    y_pred = model.predict(X_hist)

    r2 = r2_score(y_hist, y_pred)
    mae = mean_absolute_error(y_hist, y_pred)
    rmse = np.sqrt(mean_squared_error(y_hist, y_pred))
    acc = 100 - ((mae / y_hist.mean()) * 100)

    results.append({
        'Model': name,
        'R2 Score (%)': r2 * 100,
        'Accuracy (%)': acc,
        'MAE (ton)': mae,
        'RMSE (ton)': rmse
    })
    trained_models[name] = model


df_benchmark = pd.DataFrame(results).sort_values(by='R2 Score (%)', ascending=False)

print("="*60)
print("(Multi-Model Benchmark)")
print("="*60)
print(df_benchmark.to_string(index=False))

best_model_name = df_benchmark.iloc[0]['Model']
best_model = trained_models[best_model_name]

print("\n" + "="*60)
print(f": [{best_model_name}]")
print("="*60)


future_years = np.array(range(2025, 2031)).reshape(-1, 1)
co2_bau_best = best_model.predict(future_years)


renew_target_42 = np.linspace(df_egypt['renewables_share_elec'].iloc[-1], 42.0, 6)
reduction_renew = (renew_target_42 - np.linspace(df_egypt['renewables_share_elec'].iloc[-1], 20.0, 6)) * 0.008
efficiency_discount = np.linspace(0.025, 0.15, 6)
fossil_transition_discount = np.linspace(0.015, 0.10, 6)

co2_scen1_renew = co2_bau_best * (1 - reduction_renew)
co2_scen2_efficiency = co2_bau_best * (1 - efficiency_discount)
co2_scen3_fossil = co2_bau_best * (1 - fossil_transition_discount)
co2_scen4_combined = co2_bau_best * (1 - reduction_renew - efficiency_discount - fossil_transition_discount)

df_dashboard6_best = pd.DataFrame({
    'year': range(2025, 2031),
    'renewables_share_Target_%': renew_target_42,
    'CO2_BAU': co2_bau_best,
    'CO2_Scenario1_Renewables': co2_scen1_renew,
    'CO2_Scenario2_Efficiency': co2_scen2_efficiency,
    'CO2_Scenario3_FossilTransition': co2_scen3_fossil,
    'CO2_Scenario4_CombinedAction': co2_scen4_combined,
    'Avoided_CO2_BestScenario': co2_bau_best - co2_scen4_combined
})

with pd.ExcelWriter('Egypt_Dashboard6_Scenarios_BestModel.xlsx', engine='openpyxl') as writer:
    df_egypt.to_excel(writer, sheet_name='Historical_Data', index=False)
    df_dashboard6_best.to_excel(writer, sheet_name='Dashboard6_Scenarios', index=False)

print("done")
```
