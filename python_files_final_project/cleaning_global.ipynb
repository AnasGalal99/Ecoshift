```python
import pandas as pd
```

```python
df_raw = pd.read_csv('owid-energy-data.csv')
```

```python
df_filtered = df_raw[(df_raw['year'] >= 2000) & (df_raw['year'] <= 2022)].copy()
```

```python
target_columns = [
    'country', 'year', 'iso_code', 'population', 'gdp', 'primary_energy_consumption',
    'energy_per_capita', 'energy_per_gdp', 'energy_cons_change_pct', 'energy_cons_change_twh',
    'greenhouse_gas_emissions', 'carbon_intensity_elec', 'electricity_generation',
    'electricity_demand', 'electricity_demand_per_capita', 'per_capita_electricity',
    'fossil_fuel_consumption', 'fossil_electricity', 'fossil_elec_per_capita',
    'fossil_share_elec', 'fossil_share_energy', 'coal_consumption', 'coal_electricity',
    'coal_elec_per_capita', 'coal_production', 'coal_prod_per_capita', 'coal_share_elec',
    'coal_share_energy', 'oil_consumption', 'oil_electricity', 'oil_elec_per_capita',
    'oil_production', 'oil_prod_per_capita', 'oil_share_elec', 'oil_share_energy',
    'gas_consumption', 'gas_electricity', 'gas_elec_per_capita', 'gas_production',
    'gas_prod_per_capita', 'gas_share_elec', 'gas_share_energy', 'renewables_consumption',
    'renewables_electricity', 'renewables_elec_per_capita', 'renewables_share_elec',
    'renewables_share_energy', 'solar_consumption', 'solar_electricity',
    'solar_elec_per_capita', 'solar_share_elec', 'solar_share_energy', 'wind_consumption',
    'wind_electricity', 'wind_elec_per_capita', 'wind_share_elec', 'wind_share_energy',
    'hydro_consumption', 'hydro_electricity', 'hydro_elec_per_capita', 'hydro_share_elec',
    'hydro_share_energy', 'nuclear_consumption', 'nuclear_electricity',
    'nuclear_elec_per_capita', 'nuclear_share_elec', 'nuclear_share_energy',
    'biofuel_consumption', 'biofuel_electricity', 'biofuel_elec_per_capita',
    'biofuel_share_elec', 'biofuel_share_energy']
```

```python
existing_cols = [col for col in target_columns if col in df_filtered.columns]
df_cleaned = df_filtered[existing_cols].copy()
```

```python
df_cleaned.reset_index(drop=True, inplace=True)
```

```python
df_cleaned.to_csv('carbon_footprint_cleaned.csv', index=False)
```

```python
print("Data cleaned successfully!")
print("Total dataset dimensions:", df_cleaned.shape)
```

```python
df_global = pd.read_csv('carbon_footprint_cleaned.csv')
```

```python
print("Rows and columns before cleaning:", df_global.shape)
```

```python
patterns_to_remove = r'\(EI\)|\(EIA\)|\(Shift\)|\(Ember\)|^Other|^Rest of'
```

```python
df_clean_countries = df_global[~df_global['country'].str.contains(patterns_to_remove, regex=True, na=False)].copy()
```

```python
df_clean_countries = df_clean_countries[df_clean_countries['country'] != 'Antarctica']
```

```python
print("Rows and columns after cleaning country column:", df_clean_countries.shape)
print("Number of remaining unique countries:", df_clean_countries['country'].nunique())
```

```python
df_filtered_years = df_clean_countries[(df_clean_countries['year'] >= 2000) & (df_clean_countries['year'] <= 2026)].copy()
```

```python
print("Current time range:", df_filtered_years['year'].min(), "to", df_filtered_years['year'].max())
```

```python
df_final = df_filtered_years.rename(columns={'greenhouse_gas_emissions': 'annual_co2_emissions'})
```

```python
df_final.to_csv('Global_Master_Final_Clean.csv', index=False)
```

```python
print("Final table shape:", df_final.shape)
```

```python
df_final.to_csv('carbon_footprint_cleaned.csv', index=False)

print("File saved successfully as 'carbon_footprint_cleaned.csv'!")
```
