```python
import pandas as pd
```

```python
df = pd.read_excel('egypt (1).xlsx')
```

```python
df.columns = df.columns.str.strip().str.lower()
df.columns = df.columns.str.replace(' ', '_').str.replace(',', '').str.replace('â€™', '').str.replace('â€', '')
```

```python
print(df.columns.tolist()[:10])
```

```python
df.columns = df.columns.str.replace('coâ‚‚', 'co2').str.replace('co₂', 'co2')
new_order = [
    'entity', 'code', 'year',
    'annual_co2_emissions', 'co2_emissions_per_capita', 'territorial_emissions',
    'consumption-based_emissions', 'consumption_minus_territorial_co2',
    'share_of_global_annual_co2_emissions', 'cumulative_co2_emissions',
    'share_of_global_cumulative_co2_emissions',
    'per_capita_greenhouse_gas_emissions_including_land_use',
    'per_capita_methane_emissions_including_land_use',
    'per_capita_nitrous_oxide_emissions_including_land_use',
    'oil', 'gas', 'coal', 'flaring', 'cement',
    'agriculture', 'land-use_change_and_forestry', 'waste', 'buildings',
    'industry', 'factoring_and_con', 'transport',
    'gdp', 'population', 'carbon_intensity_of_energy',
    'total_energy_supply_per_unit_of_gdp', 'primary_energy_consumption',
    'solar_electricity', 'solar_share_elec', 'wind_electricity',
    'wind_share_elec', 'fossil_share_elec', 'renewables_share_elec'
]
existing_cols = [col for col in new_order if col in df.columns]
df = df[existing_cols]
```

```python
print(df.columns.tolist()[:10])
```

```python
df = df[(df['year'] >= 2000) & (df['year'] <= 2024)].reset_index(drop=True)
```

```python
missing_data = df.isnull().sum()
print(missing_data[missing_data > 0])
```

```python
df = df.ffill()
print("Total remaining null values in the dataset:", df.isnull().sum().sum())
```

```python
df['year'] = df['year'].astype(int)
```

```python
df['co2_growth_rate_%'] = df['annual_co2_emissions'].pct_change() * 100
```

```python
df['fossil_to_renewable_ratio'] = df['fossil_share_elec'] / (df['renewables_share_elec'] + 0.001)
```

```python
df['co2_growth_rate_%'] = df['co2_growth_rate_%'].fillna(0)
```

```python
print(df[['year', 'annual_co2_emissions', 'co2_growth_rate_%', 'fossil_to_renewable_ratio']].head())
```

```python
print("Total null values in the file:", df.isnull().sum().sum())
```

```python
print(f"Time range: from {df['year'].min()} to {df['year'].max()}")
```

```python
df.to_csv("Egypt_Master_Clean.csv", index=False)
print("Master dataset saved successfully as: Egypt_Master_Clean.csv")
```
